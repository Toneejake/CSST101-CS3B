# Part 1: Implement Basic Probabilty Calculations

1. Task: Write Python functions to calculate the following:

Joint Probability

```python
def joint_probability(p_A, p_B):
  return p_A * p_B
```

Marginal Probability

```python
def marginal_probability(p_A, p_B):
  return p_A + p_B - joint_probability(p_A, p_B)
```

Conditional Probality

```python
def conditional_probability(p_B_given_A, p_A, p_B):
  return (p_B_given_A * p_A) / p_B
```

2. Interactive Component:

```python
p_A = float(input("Enter probability of event A: "))
p_B = float(input("Enter probability of event B: "))
p_B_given_A = float(input("Enter probability of B given A: "))

print("\nJoint Probability:", joint_probability(p_A, p_B))
print("Marginal Probability:", marginal_probability(p_A, p_B))
print("Conditional Probability:", conditional_probability(p_B_given_A, p_A, p_B))
```

# Part 2: Bayesian Inference for Real-World Scenarios

1. Task: Implement a Python function that uses Bayesian inference to update
probabilities based on new evidence.

```python
def bayesian_inference(prior, likelihood_positive_given_disease, likelihood_positive):
  posterior = (likelihood_positive_given_disease * prior) / likelihood_positive
  return posterior
```

2. Interactive Component:

```python
prior_disease = float(input("Enter prior probability of disease: "))
likelihood_positive_given_disease = float(input("Enter likelihood of positive test given disease: "))
likelihood_positive = float(input("Enter likelihood of positive test: "))

posterior_probability = bayesian_inference(prior_disease, likelihood_positive_given_disease, likelihood_positive)

print("\nPosterior Probability:", posterior_probability)
```

# Part 3: Simulate Decision-Making Under Uncertainty

1. Task: Write a Python function to simulate a decision-making process where outcomes are uncertain (e.g., investment decisions). Use a probability distribution to model the decision’s potential outcomes.


```python
def simulate_decision(num_simulations, p_success, reward_success, reward_failure):
  outcomes = []
  for _ in range(num_simulations):
    if np.random.rand() < p_success:
      outcomes.append(reward_success)
    else:
      outcomes.append(reward_failure)
  return np.mean(outcomes)
```

2. Interactive Component:

```python
p_success = float(input("Enter probability of success: "))
reward_success = float(input("Enter reward for success: "))
reward_failure = float(input("Enter penalty for failure: "))

average_outcome = simulate_decision(1000, p_success, reward_success, reward_failure)
print(f"Expected value of decision: {average_outcome}")
```

# Part 4: Visualize Probability Distributions

1. Task: Use Python to generate and visualize binomial and normal probability
distributions. The visualizations should represent real-world scenarios like coin
flips or dice rolls.

2. Interactive Component:

```python
print("Choose a distribution to simulate:")
print("1. Binomial (Coin Flip)")
print("2. Normal (Exam Scores)")

choice = int(input("Enter 1 or 2: "))

if choice == 1:
    print("\nSimulating Binomial Distribution (Coin Flip)")
    trials = int(input("Enter the number of trials: "))
    prob = float(input("Enter the probability of heads(for binomial): "))

    n = 1
    binomial_data = np.random.binomial(trials, prob, n)

    plt.figure(figsize=(8, 6))
    plt.hist(binomial_data, bins=2, edgecolor='black')
    plt.title('Binomial Distribution (Coin Flips - Number of Heads)')
    plt.xlabel('Number of Heads')
    plt.ylabel('Frequency')
    plt.xticks(np.arange(binomial_data.min(), binomial_data.max()+1))
    plt.show()

elif choice == 2:
    print("\nSimulating Normal Distribution (Exam Scores)")
    mean = float(input("Enter the mean (e.g., average exam score): "))
    std = float(input("Enter the standard deviation (e.g., score variability): "))

    n = 1
    normal_data = np.random.normal(mean, std, n)

    plt.figure(figsize=(8, 6))
    plt.hist(normal_data, bins=30, edgecolor='black')
    plt.title('Normal Distribution (Exam Scores)')
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.show()

else:
    print("Invalid choice.")
```

# Part 5: Real-World Scenario Prediction

1. Task: Write a Python script that uses conditional probability to predict a realworld event. The scenario should involve predicting the probability of rain based on
inputs like humidity and cloud cover.

```python
def predict_rain(rain, humidity, cloud_cover):
  return (rain * humidity) / cloud_cover
```

```python
humidity = float(input("Enter the humidity(0 to 1): "))
cloud_cover = float(input("Enter the cloud cover(0 to 1): "))
rain = 0.3

probability_of_rain = predict_rain(rain, humidity, cloud_cover)
print(f"Probability of rain: {probability_of_rain}")
```
