Part 4: Visualize a Probabilty Distribution

```python
coin_flips = 5
p_head = 0.5
binomial_distribution = np.random.binomial(n=1, p=p_head, size=coin_flips)
print(binomial_distribution)
```

```python
x = np.arange(0, coin_flips + 1)
pmf = binom.pmf(x, coin_flips, p_head)

plt.bar(x, pmf)
plt.xlabel('Number of Heads')
plt.ylabel('Probability')
plt.title('Probability Distribution of Getting Heads in 5 Coin Flips')
plt.show()
```
